## ③ Coherencia interna del paper — parser estructurado (semilla B1)

El paper fuente usa entornos LaTeX (`\begin{theorem}[...]`) → extracción
**determinística**, sin OCR/PDF. Este parser es la base del nivel L2.


````python
tex = main_tex.read_text(errors="replace")
env_re = re.compile(r"\\begin\{(theorem|lemma|corollary|proposition|definition|remark)\}(?:\[([^\]]+)\])?")
rows = []
for m in env_re.finditer(tex):
    line = tex.count("\n", 0, m.start()) + 1
    rows.append((m.group(1), m.group(2) or "", line))
print(f"Entornos LaTeX en main.tex: {len(rows)}")
for kind, name, ln in rows:
    print(f"  L{ln:5d}  {kind:<11} {name}")
````

17 entornos: **2 theorems** (Hybrid invariance L729 · Convergence in expectation L824 —
conjetura verificada contra la extracción del PDF), 5 definitions, 8 remarks,
1 corollary (Lyapunov–Landauer L2526), 1 proposition (S5 L4205).
