## ls20 — Key Rotation

### Grid
- Size: 64x64
- Colors: 10 (0-9)

### Zonas
| Zona | Rows | Cols | Centro | Descripción |
|---|---|---|---|---|
| Model (key) | 53-62 | 0-15 | [56, 5] | Figura que rota |
| Figure (lock) | 9-15 | 14-26 | [12, 20] | Destino final |
| Checkpoint | 31-33 | 0-63 | [32, 31] | Rota la key |
| Life bar | 60-62 | 0-63 | [61, 31] | Presupuesto |
| Player | 8 | 32 | [8, 32] | Posición inicial |

### Mecánica
1. Player navega laberinto
2. Key rota cuando player pasa por checkpoint
3. Key calza cuando patrón == figure (colores idénticos)
4. Gana cuando llega a figure con key correcta

### Invariantes
- `life_decreases_by_2`: cada movimiento cuesta 2
- `model_shape_constant`: forma de key no cambia
- `figure_shape_constant`: forma de figure no cambia
- `key_fits_figure`: detectado por comparación de colores

### Estrategia Óptima
1. Calcular distancia player → checkpoint
2. Calcular distancia checkpoint → figure
3. Verificar si vida alcanza (distancia_total * 2)
4. Navegar a checkpoint (rotar key)
5. Navegar a figure (ganar)

- ✅ Zone detection implementado
- ✅ Key-fit detection implementado
- ❌ Pathfinding no implementado
- ❌ Resuelto: 0/1
